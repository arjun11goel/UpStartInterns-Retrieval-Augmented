# Task 11: NumPy Student Marks
**Objective:** Learn how NumPy can efficiently perform numerical analysis[cite: 1].

**Description:** Create a NumPy array containing student marks and calculate the mean, median, maximum, minimum, and standard deviation[cite: 1].

```python
import numpy as np

# Manually created student marks out of 100[cite: 1]
student_marks = np.array([85, 78, 92, 65, 88, 72, 95, 55, 81, 90])

print("Array of Student Marks:")
print(student_marks)
```

### Statistical Analysis
Using NumPy's built-in functions (`np.mean()`, `np.median()`, `np.max()`, `np.min()`, and `np.std()`) to analyze the dataset[cite: 1].

```python
# Calculating statistical values[cite: 1]
mean_mark = np.mean(student_marks)
median_mark = np.median(student_marks)
max_mark = np.max(student_marks)
min_mark = np.min(student_marks)
std_dev = np.std(student_marks)

# Display each result clearly[cite: 1]
print("--- Statistical Analysis Results ---")
print(f"Mean (Average): {mean_mark}")
print(f"Median: {median_mark}")
print(f"Maximum Mark: {max_mark}")
print(f"Minimum Mark: {min_mark}")
print(f"Standard Deviation: {std_dev:.2f}")
```

### Explanation of the Results
* **Mean (80.1):** The average score of all students in the dataset is roughly 80.
* **Median (83.0):** When all marks are ordered from lowest to highest, the middle value is 83. Because this is higher than the mean, the lower scores (like 55 and 65) pull the average down slightly.
* **Maximum (95) & Minimum (55):** These represent the range of the class performance, showing the highest and lowest scores achieved.
* **Standard Deviation (12.28):** On average, the students' marks vary by about 12.28 points from the mean. A larger standard deviation means the scores are more spread out, whereas a lower one would mean most students scored very close to 80.1.

### Interview Questions & Answers[cite: 1]

**1. What is NumPy?**
NumPy (Numerical Python) is a foundational Python library used for scientific computing. It provides support for large, multi-dimensional arrays and matrices, along with a vast collection of high-level mathematical functions to operate on these arrays efficiently.

**2. Why is NumPy useful in machine learning?**
Machine learning algorithms rely heavily on linear algebra, matrix multiplications, and statistical operations. NumPy performs these computations efficiently because its core operations are written in C, allowing for "vectorization" (processing entire arrays at once rather than looping through individual elements in standard Python lists).

**3. What is standard deviation?**
Standard deviation is a statistical metric that measures the dispersion or spread of a dataset relative to its mean. A low standard deviation indicates that the data points tend to be very close to the mean, while a high standard deviation indicates that the data points are spread out over a larger range of values.
